```python
import pandas as pd
import os

DATASET_PATH = "/kaggle/input/datasets/awsaf49/cbis-ddsm-breast-cancer-image-dataset"
CSV_PATH = os.path.join(DATASET_PATH, "csv")

mass_train = pd.read_csv(
    os.path.join(CSV_PATH, "mass_case_description_train_set.csv")
)

calc_train = pd.read_csv(
    os.path.join(CSV_PATH, "calc_case_description_train_set.csv")
)

print("MASS")
print("=" * 60)

print("Image path:")
print(mass_train["image file path"].iloc[0])

print("\nCropped image path:")
print(mass_train["cropped image file path"].iloc[0])

print("\nROI mask path:")
print(mass_train["ROI mask file path"].iloc[0])


print("\n\nCALCIFICATION")
print("=" * 60)

print("Image path:")
print(calc_train["image file path"].iloc[0])

print("\nCropped image path:")
print(calc_train["cropped image file path"].iloc[0])

print("\nROI mask path:")
print(calc_train["ROI mask file path"].iloc[0])
```

```python
# Check DICOM information and image path mapping

DICOM_INFO_PATH = os.path.join(CSV_PATH, "dicom_info.csv")

dicom_info = pd.read_csv(DICOM_INFO_PATH)

print("Shape:", dicom_info.shape)

print("\nFirst 3 file_path:")
print(dicom_info["file_path"].head(3).to_string(index=False))

print("\nFirst 3 image_path:")
print(dicom_info["image_path"].head(3).to_string(index=False))
```

```python
# ============================================================
# Find JPEG path using the DICOM unique identifier
# ============================================================

mass_path = mass_train["image file path"].iloc[0]

# Get the DICOM unique ID from the case path
parts = mass_path.split("/")
dicom_uid = parts[-2]

print("DICOM UID:")
print(dicom_uid)

# Search for this UID in dicom_info file_path
match = dicom_info[
    dicom_info["file_path"].str.contains(dicom_uid, na=False)
]

print("\nNumber of matches:", len(match))

if len(match) > 0:
    print("\nMatched DICOM path:")
    print(match["file_path"].iloc[0])

    print("\nMatched JPEG path:")
    print(match["image_path"].iloc[0])
else:
    print("\nNo match found.")
```

```python
import matplotlib.pyplot as plt
from PIL import Image

# Get JPEG path from dicom_info
jpeg_relative_path = match["image_path"].iloc[0]

# Remove "CBIS-DDSM/" from the beginning
jpeg_relative_path = jpeg_relative_path.replace("CBIS-DDSM/", "")

# Create actual Kaggle path
jpeg_full_path = os.path.join(DATASET_PATH, jpeg_relative_path)

print("JPEG path:")
print(jpeg_full_path)

print("\nFile exists:", os.path.exists(jpeg_full_path))

# Load and display image
image = Image.open(jpeg_full_path)

print("Image size:", image.size)
print("Image mode:", image.mode)

plt.figure(figsize=(8, 8))
plt.imshow(image, cmap="gray")
plt.axis("off")
plt.title("CBIS-DDSM Mammogram")
plt.show()
```

```python
# Show metadata for the same Mass case

case = mass_train.iloc[0]

print("Patient ID       :", case["patient_id"])
print("Breast density   :", case["breast_density"])
print("Breast side      :", case["left or right breast"])
print("Image view       :", case["image view"])
print("Abnormality type :", case["abnormality type"])
print("Pathology        :", case["pathology"])
print("Assessment       :", case["assessment"])
print("Subtlety         :", case["subtlety"])
```

```python
def extract_uid(path):
    parts = path.split("/")
    return parts[-2]


# Create UID → JPEG path mapping
dicom_info["uid"] = dicom_info["file_path"].apply(extract_uid)

uid_to_jpeg = dict(
    zip(
        dicom_info["uid"],
        dicom_info["image_path"]
    )
)

print("Total UID mappings:", len(uid_to_jpeg))
```

```python
mass_train["uid"] = mass_train["image file path"].apply(extract_uid)

mass_train["jpeg_path"] = mass_train["uid"].map(uid_to_jpeg)

mass_train["jpeg_path"] = mass_train["jpeg_path"].apply(
    lambda x: os.path.join(
        DATASET_PATH,
        x.replace("CBIS-DDSM/", "")
    ) if pd.notna(x) else None
)

print("Mass cases:", len(mass_train))
print("JPEG found:", mass_train["jpeg_path"].notna().sum())

calc_train["uid"] = calc_train["image file path"].apply(extract_uid)

calc_train["jpeg_path"] = calc_train["uid"].map(uid_to_jpeg)

calc_train["jpeg_path"] = calc_train["jpeg_path"].apply(
    lambda x: os.path.join(
        DATASET_PATH,
        x.replace("CBIS-DDSM/", "")
    ) if pd.notna(x) else None
)

print("Calcification cases:", len(calc_train))
print("JPEG found:", calc_train["jpeg_path"].notna().sum())
```

```python
mass_clean = pd.DataFrame({
    "patient_id": mass_train["patient_id"],
    "image_path": mass_train["jpeg_path"],
    "density": mass_train["breast_density"],
    "breast_side": mass_train["left or right breast"],
    "image_view": mass_train["image view"],
    "abnormality_type": mass_train["abnormality type"],
    "pathology": mass_train["pathology"],
    "roi_path": mass_train["ROI mask file path"]
})

calc_clean = pd.DataFrame({
    "patient_id": calc_train["patient_id"],
    "image_path": calc_train["jpeg_path"],
    "density": calc_train["breast density"],
    "breast_side": calc_train["left or right breast"],
    "image_view": calc_train["image view"],
    "abnormality_type": calc_train["abnormality type"],
    "pathology": calc_train["pathology"],
    "roi_path": calc_train["ROI mask file path"]
})

print("Mass:", mass_clean.shape)
print("Calcification:", calc_clean.shape)

train_df = pd.concat(
    [mass_clean, calc_clean],
    ignore_index=True
)

print("Total cases:", len(train_df))
print(train_df.head())

```

```python
train_df["label"] = train_df["pathology"].map({
    "MALIGNANT": 1,
    "BENIGN": 0,
    "BENIGN_WITHOUT_CALLBACK": 0
})

train_df["density"] = train_df["density"].astype(int)

print("Label distribution:")
print(train_df["label"].value_counts())

print("\nDensity distribution:")
print(train_df["density"].value_counts().sort_index())
```

```python
def extract_roi_uid(path):
    parts = path.split("/")
    return parts[-2]


mass_clean["roi_uid"] = mass_train["ROI mask file path"].apply(
    extract_roi_uid
)

calc_clean["roi_uid"] = calc_train["ROI mask file path"].apply(
    extract_roi_uid
)

print("Mass ROI UID:", mass_clean["roi_uid"].iloc[0])
print("Calc ROI UID:", calc_clean["roi_uid"].iloc[0])
```

```python
roi_uid_to_jpeg = dict(
    zip(
        dicom_info["uid"],
        dicom_info["image_path"]
    )
)
mass_clean["roi_jpeg_path"] = mass_clean["roi_uid"].map(
    roi_uid_to_jpeg
)

calc_clean["roi_jpeg_path"] = calc_clean["roi_uid"].map(
    roi_uid_to_jpeg
)

print(
    "Mass ROI found:",
    mass_clean["roi_jpeg_path"].notna().sum(),
    "/",
    len(mass_clean)
)

print(
    "Calcification ROI found:",
    calc_clean["roi_jpeg_path"].notna().sum(),
    "/",
    len(calc_clean)
)
```

```python
missing_calc = calc_clean[
    calc_clean["roi_jpeg_path"].isna()
]

print(missing_calc[
    ["patient_id", "roi_uid"]
])

missing_uid = missing_calc["roi_uid"].iloc[0]

print("Missing ROI UID:", missing_uid)

print(
    calc_train[
        calc_train["patient_id"] == missing_calc["patient_id"].iloc[0]
    ][[
        "patient_id",
        "ROI mask file path"
    ]]
)
```

```python
missing_patient = missing_calc["patient_id"].iloc[0]

missing_case = calc_train[
    (calc_train["patient_id"] == missing_patient) &
    (calc_train["ROI mask file path"].str.contains(missing_uid))
]

print(missing_case[
    [
        "patient_id",
        "image file path",
        "cropped image file path",
        "ROI mask file path",
        "pathology",
        "breast density",
        "left or right breast",
        "image view"
    ]
])
```

```python
original_path = missing_case["image file path"].iloc[0]

original_uid = extract_uid(original_path)

print("Original image UID:")
print(original_uid)

print("\nJPEG mapping exists:",
      original_uid in uid_to_jpeg)
```

```python
mass_clean["roi_path"] = mass_clean["roi_jpeg_path"].apply(
    lambda x: os.path.join(
        DATASET_PATH,
        x.replace("CBIS-DDSM/", "")
    ) if pd.notna(x) else None
)

calc_clean["roi_path"] = calc_clean["roi_jpeg_path"].apply(
    lambda x: os.path.join(
        DATASET_PATH,
        x.replace("CBIS-DDSM/", "")
    ) if pd.notna(x) else None
)
final_df = pd.concat(
    [mass_clean, calc_clean],
    ignore_index=True
)

print("Total cases:", len(final_df))
print("ROI available:", final_df["roi_path"].notna().sum())
print("ROI missing:", final_df["roi_path"].isna().sum())
```

```python
final_df["label"] = final_df["pathology"].map({
    "MALIGNANT": 1,
    "BENIGN": 0,
    "BENIGN_WITHOUT_CALLBACK": 0
})

final_df["density"] = final_df["density"].astype(int)
print(final_df.columns.tolist())
print(final_df.shape)
```

```python
print("Unique patients:", final_df["patient_id"].nunique())
print("Total rows:", len(final_df))
patient_counts = final_df["patient_id"].value_counts()

print(patient_counts.head(10))
print(
    "Patients with multiple rows:",
    (patient_counts > 1).sum()
)
```

```python
from sklearn.model_selection import train_test_split
patient_ids = final_df["patient_id"].unique()

train_patients, val_patients = train_test_split(
    patient_ids,
    test_size=0.20,
    random_state=42
)

print("Train patients:", len(train_patients))
print("Validation patients:", len(val_patients))
train_df = final_df[
    final_df["patient_id"].isin(train_patients)
].copy()

val_df = final_df[
    final_df["patient_id"].isin(val_patients)
].copy()

print("Train rows:", len(train_df))
print("Validation rows:", len(val_df))
```

```python
overlap = set(train_df["patient_id"]) & set(val_df["patient_id"])

print("Patient overlap:", len(overlap))
```

```python
print("Train labels:")
print(train_df["label"].value_counts())

print("\nValidation labels:")
print(val_df["label"].value_counts())

print("\nTrain density:")
print(train_df["density"].value_counts().sort_index())

print("\nValidation density:")
print(val_df["density"].value_counts().sort_index())
```
